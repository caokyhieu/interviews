```python
class Trade:

    def __init__(self, symbol, quantity, price):
        self.symbol = symbol
        self.quantity = quantity
        self.price = price

    @property
    def notional(self):
        return self.quantity * self.price
    
    def __str__(self):
        return f"Trade(symbol={self.symbol}, quantity={self.quantity}, price={self.price}"
    
    def __repr__(self):
        return f"Trade(symbol={self.symbol}, quantity={self.quantity}, price={self.price})"
    
    def __eq__(self, other):
        if isinstance(other, Trade):
            return (self.symbol == other.symbol and
                    self.quantity == other.quantity and
                    self.price == other.price)
        return False
    
    def __hash__(self):
        return hash((self.symbol, self.quantity, self.price))
    

class Porfolio:

    def __init__(self):
        self.trades = []

    def add_trade(self, trade):
        if isinstance(trade, Trade):
            self.trades.append(trade)
        else:
            raise ValueError("Only Trade instances can be added to the portfolio.")

    def __iter__(self):
        yield from self.trades
        
    def total_notional(self):
        return sum(trade.notional for trade in self.trades)

    def __str__(self):
        return f"Portfolio with {len(self.trades)} trades, Total Notional: {self.total_notional()}"
    
    def __repr__(self):
        return f"Portfolio(trades={self.trades})"
```

```python
class Node:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def inorder_traversal(root):
    if not root:
        return []
    return inorder_traversal(root.left) + [root.val] + inorder_traversal(root.right)

## test cases
root = Node(1, Node(2, Node(4), Node(5)), Node(3))
print(inorder_traversal(root))
```

```python
## use stack to implement inorder traversal
def inorder_traversal_stack(root):
    stack = []
    result = []
    curr = root

    while curr is not None or stack:
        # go as left as possible
        while curr is not None:
            stack.append(curr)
            curr = curr.left

        # now curr is None, pop from stack
        curr = stack.pop()
        result.append(curr.val)   # visit node
        curr = curr.right         # move to right subtree

    return result

## test cases
root = Node(1, Node(2, Node(4), Node(5)), Node(3))
print(inorder_traversal_stack(root))
```

```python
def preorder_traversal(root):
    if not root:
        return []
    return [root.val] + preorder_traversal(root.left) + preorder_traversal(root.right)

## test
root = Node(1, Node(2, Node(4), Node(5)), Node(3))
print(preorder_traversal(root))
```

```python
def preorder_traversal_stack(root):
    if not root:
        return []
    
    stack = [root]
    result = []

    while stack:
        node = stack.pop()
        result.append(node.val)

        # push right first so that left is processed first
        if node.right:
            stack.append(node.right)
        if node.left:
            stack.append(node.left)

    return result


## test
root = Node(1, Node(2, Node(4), Node(5)), Node(3))
print(preorder_traversal_stack(root))
```

```python
def breadth_first_traversal(root):
    if not root:
        return []
    
    from collections import deque
    queue = deque([root])
    result = []

    while queue:
        node = queue.popleft()
        result.append(node.val)

        if node.left:
            queue.append(node.left)
        if node.right:
            queue.append(node.right)

    return result

## test
root = Node(1, Node(2, Node(4), Node(5)), Node(3))
print(breadth_first_traversal(root))
```

```python
def subsets(nums):
    res = []
    path = []

    def backtrack(i):
        if i == len(nums):
            res.append(path.copy())
            return
        # choice: skip nums[i]
        backtrack(i + 1)
        # choice: take nums[i]
        path.append(nums[i])
        backtrack(i + 1)
        path.pop()

    backtrack(0)
    return res
```
